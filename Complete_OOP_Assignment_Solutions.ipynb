### Question 1: Classes and Objects (Beginner) - Smart Home Hub
**Scenario:** In the Smart Home Hub project, the team needs to represent the core entity: `Device`.
**Task:** Create a basic `Device` class initializing with `id` and `temperature`. Add a method `trigger()` that prints a status message.


```python
class Device:
    def __init__(self, id, temperature):
        self.id = id
        self.temperature = temperature

    def trigger(self):
        print(f"[Device {self.id}] Device triggered and functioning.")

# Test Question 1
obj_1 = Device(101, 50)
obj_1.trigger()
```

### Question 2: Encapsulation & Properties (Intermediate) - Smart Home Hub
**Scenario:** Security audits revealed `temperature` was being altered maliciously or accidentally.
**Task:** Modify `Device` to make `temperature` private (`_temperature`). Provide a property getter and setter that raises `ValueError` if negative.


```python
class DeviceEncapsulated:
    def __init__(self, id, temperature):
        self.id = id
        self.temperature = temperature  # Triggers the property setter

    @property
    def temperature(self):
        return self._temperature

    @temperature.setter
    def temperature(self, value):
        if value < 0:
            raise ValueError(f"'temperature' cannot be negative: {value}")
        self._temperature = value

# Test Question 2
enc_2 = DeviceEncapsulated(102, 75)
print(f"Valid temperature: {enc_2.temperature}")
try:
    enc_2.temperature = -10
except ValueError as e:
    print(f"Caught expected error: {e}")
```

### Question 3: Inheritance (Intermediate) - Smart Home Hub
**Scenario:** The platform is expanding. Standard `Device` objects aren't enough. We need specialized behaviors for `Thermostat`.
**Task:** Create subclass `Thermostat` that inherits from `Device`. Add attribute `special_flag` and override `__init__` using `super()`.


```python
class Thermostat(Device):
    def __init__(self, id, temperature, special_flag):
        super().__init__(id, temperature)
        self.special_flag = special_flag

# Test Question 3
sub_obj_3 = Thermostat(103, 80, special_flag=True)
print(f"Created Thermostat: ID={sub_obj_3.id}, temperature={sub_obj_3.temperature}, special_flag={sub_obj_3.special_flag}")
```

### Question 4: Polymorphism (Industry) - Smart Home Hub
**Scenario:** An orchestrator module needs to process a list of different `Device` types seamlessly.
**Task:** Override `trigger()` in `Thermostat` to implement custom behavior. Write `process_all(entities)` that accepts a list and triggers `trigger()` on all.


```python
class ThermostatPolymorphic(Device):
    def __init__(self, id, temperature, special_flag):
        super().__init__(id, temperature)
        self.special_flag = special_flag

    def trigger(self):
        print(f"[Thermostat {self.id} | Flag: {self.special_flag}] Thermostat adjusting temperature regulation.")

def process_all_smart_home_hub(entities):
    for entity in entities:
        entity.trigger()

# Test Question 4
sample_list_4 = [
    Device(101, 40),
    ThermostatPolymorphic(102, 60, special_flag=True)
]
process_all_smart_home_hub(sample_list_4)
```

### Question 5: Magic Methods (Industry) - Smart Home Hub
**Scenario:** Logs are cluttered with memory addresses. Devops needs readable logging, and analysts want to compare performance based on `temperature`.
**Task:** Implement `__str__` for `Device` to return a readable string, and `__gt__` to compare two objects based on `temperature`.


```python
class DeviceComplete(Device):
    def __str__(self):
        return f"Device(id={self.id}, temperature={self.temperature})"

    def __gt__(self, other):
        if not isinstance(other, Device):
            return NotImplemented
        return self.temperature > other.temperature

# Test Question 5
item_a_5 = DeviceComplete(1, 45)
item_b_5 = DeviceComplete(2, 85)

print("Readable String representation:", str(item_a_5))
print(f"Is Item B > Item A? {item_b_5 > item_a_5}")
```

### Challenge 4: Class & Static Methods
**Challenge Focus:** Smart Home API Parsing & MAC Validation (Instance 1)


```python
# Challenge 4 Solution
class Device:
    def __init__(self, mac_address, temperature):
        self.mac_address = mac_address
        self.temperature = float(temperature)

    @staticmethod
    def validate_mac_address(mac_str):
        return len(str(mac_str)) > 3

    @classmethod
    def from_string(cls, raw_str):
        # String formatted as 'attr2-attr1' -> 'temperature-mac_address'
        parts = raw_str.split("-", 1)
        attr2 = parts[0]
        attr1 = parts[1]
        return cls(mac_address=attr1, temperature=attr2)

# Verification
valid = Device.validate_mac_address("00:1A:2B")
print(f"MAC Valid? {valid}")
dev = Device.from_string("72.5-AA:BB:CC:DD")
print(f"Parsed Device -> MAC: {dev.mac_address}, Temp: {dev.temperature}°F")
```

### Challenge 5: Magic Methods
**Challenge Focus:** Logistics Shipment Aggregation & Comparison (Instance 1)


```python
# Challenge 5 Solution
class Shipment:
    def __init__(self, destination, weight):
        self.destination = destination
        self.weight = weight

    def __str__(self):
        return f"Shipment: [{self.weight}]"

    def __add__(self, other):
        if isinstance(other, Shipment):
            return self.weight + other.weight
        return self.weight + other

    def __gt__(self, other):
        if isinstance(other, Shipment):
            return self.weight > other.weight
        return self.weight > other

# Verification
s1 = Shipment("New York", 150)
s2 = Shipment("London", 275)
print(str(s1))
print(f"Total Weight: {s1 + s2} kg")
print(f"Is s2 heavier than s1? {s2 > s1}")
```

### Challenge 8: Abstract Base Classes
**Challenge Focus:** Gaming Architecture Damage Protocol (Instance 1)


```python
# Challenge 8 Solution
from abc import ABC, abstractmethod

class Character(ABC):
    def __init__(self, name, health=100):
        self.name = name
        self.health = health

    @abstractmethod
    def take_damage(self, amount):
        pass

class BossCharacter(Character):
    def __init__(self, name, health=500, shield=50):
        super().__init__(name, health)
        self.shield = shield

    def take_damage(self, amount):
        effective_damage = max(0, amount - self.shield)
        self.health -= effective_damage
        return f"{self.name} took {effective_damage} damage (shield absorbed {min(amount, self.shield)}). Health remaining: {self.health}"

# Verification
boss = BossCharacter("Malakor the Destroyer", 1000, shield=40)
print(boss.take_damage(100))
```

```python
# Challenge 49 Solution
class Server:
    def __init__(self, ip_address, cpu_load):
        self.ip_address = ip_address
        self.cpu_load = float(cpu_load)

    @staticmethod
    def validate_ip_address(ip_str):
        return len(str(ip_str)) > 3

    @classmethod
    def from_string(cls, raw_str):
        # Formatted as 'attr2-attr1' -> 'cpu_load-ip_address'
        parts = raw_str.split("-", 1)
        attr2 = parts[0]
        attr1 = parts[1]
        return cls(ip_address=attr1, cpu_load=attr2)

# Verification
print(f"Valid IP? {Server.validate_ip_address('192.168.1.1')}")
srv = Server.from_string("84.2-10.0.0.1")
print(f"Server IP: {srv.ip_address}, CPU Load: {srv.cpu_load}%")
```

### Challenge 50: Magic Methods
**Challenge Focus:** Travel Flight Aggregation & Seat Availability (Instance 5)


```python
# Challenge 50 Solution
class Flight:
    def __init__(self, flight_number, seats_available):
        self.flight_number = flight_number
        self.seats_available = seats_available

    def __str__(self):
        return f"Flight: [{self.seats_available}]"

    def __add__(self, other):
        if isinstance(other, Flight):
            return self.seats_available + other.seats_available
        return self.seats_available + other

    def __gt__(self, other):
        if isinstance(other, Flight):
            return self.seats_available > other.seats_available
        return self.seats_available > other

# Verification
f1 = Flight("AI-101", 45)
f2 = Flight("BA-202", 120)
print(str(f1))
print(f"Combined Available Seats: {f1 + f2}")
print(f"Does f2 have more seats than f1? {f2 > f1}")
```
